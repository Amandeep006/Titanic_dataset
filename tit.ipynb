```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Fetch the data of titianic 
ti=pd.read_csv("titanic.csv")
ti.head()
```

```python
# To see the overall information about the dataset.
ti.info()
```

```python
# To check the statistical information about the dataset.
ti.describe()
```

```python
# To check how many null values are present in the dataset.
ti.isnull().sum()
```

```python
# To check how many duplicates values are present in the dataset.
ti.duplicated().sum()
```

First we handle the null values present in the dataset.

```python
# Handle the null value of age column.
med=ti["age"].median()
ti["age"]=ti["age"].fillna(med)
```

```python
# Now convert the datatype of the  age column 
ti["age"]=ti["age"].astype(int)
ti.head()
```

```python
# Extract the surname and title of the  name column 

"""surname"""
ti["surname"]=ti["name"].str.split(",").str[0]

"""title"""
ti["title"]=ti["name"].str.split(",").str[1].str.split(".").str[0]
```

```python
ti["fare"]=round(ti["fare"],2)
```

# Data Analysis 

    
    How many passengers were aboard the Titanic?

```python
# Total number of passengers.
ti["pclass"].count()
```

```python
# Number of survivors.
ti[ti["survived"]==1]["survived"].count()
```

```python
# Number of non-survivors.
ti[ti["survived"]==0]["survived"].count()
```


    What was the overall survival rate?

```python
surv=ti[ti["survived"]==1]["survived"].count()
total_pass=ti["survived"].count()

percentage=(surv/total_pass)*100

print(f"Percentage of overall survial rate : {round(percentage,2)}")
```


    What was the distribution of passengers by sex?

```python
# Number of male passengers.
ma=ti[ti["sex"]=="male"]["sex"].count()
print(f"Number of male passenger : {ma}")
```

```python
# Number of female passengers.
fe=ti[ti["sex"]=="female"]["sex"].count()
print(f"Number of female passenger : {fe}")
```

```python
# Percentage of each group.

male_percent=(ma/total_pass)*100
female_percent=(fe/total_pass)*100

# display(round(male_percent,2))
# display(round(female_percent,2))

print(f"Male percentage : {round(male_percent,2)}")
print(f"female percentage : {round(female_percent,2)}")
```

```python
# Minimum and maximum fares.  
print(f"Minimum fare : {round(ti["fare"].min(),2)}")
print(f"Maximum fare : {round(ti["fare"].max(),2)}")
```


    Did survival rates differ between male and female passengers?

```python
# Male survival rate.
sur_mal=ti[(ti["sex"]=="male")&(ti["survived"]==1)]["survived"].count()
total_surv=ti[ti["survived"]==1]["survived"].count()

percent=(sur_mal/total_surv)*100

print(f"Male survival rate : {percent}")
```

```python
# Female survival rate.
sur_fem=ti[(ti["sex"]=="female")&(ti["survived"]==1)]["survived"].count()
total_surv=ti[ti["survived"]==1]["survived"].count()

percent=(sur_fem/total_surv)*100
print(f"Female survival rate : {round(percent,2)} ")
```
